## Question 1.1
rubric={reasoning:60}
    
<div class="alert alert-info">

<ol type="1">
<li>What is a data broker?</li>
<li>What are direct and indirect identifiers?</li>
<li>What is k-anonymity and l-diversity?</li>
<li>What are some weaknesses of k-anonymity (and l-diversity)? How could you re-identify individuals in these datasets?</li>
<li>If you have two differentially private datasets, one with and one without your data, what does differential privacy guarantee regarding your privacy?</li>
<li>How does differential privacy work on a conceptual level (watching the optional video from Minute Physics can help with this if you want more details and examples than what was given in the lecture)?</li>
<li>What are some additional approaches you could take to secure data stored in your organization?</li>
</ol>

</div>

YOUR ANSWERS HERE

1. A data broker is basically a company that collects personal info about people from a bunch of different sources like public records, purchase history, and online activity, then sells or licenses that data to other businesses. They kind of act as middlemen in the personal data economy, building up detailed profiles on people who usually dont even know their information is being bought and sold.

2. Direct identifiers are things that can identify a person on their own, like a full name, social security number, or email address. Indirect identifiers (or quasi-identifiers) are things like zip code, date of birth, and gender that dont identify anyone alone but when combined can single out a specific individual pretty easily.

3. K-anonymity is a privacy technique where each record in a dataset is made indistinguishable from atleast k-1 other records based on their quasi-identifiers, meaning any given combination of those attributes applies to atleast k people. L-diversity builds on this by also requiring that within each k-anonymous group, the sensitive attribute has atleast l different values so you still cant determine someones sensitive info even if you narrow them down to a group.

4. K-anonymity is vulnerable to homogeneity attacks where all records in a group share the same sensitive value, making that value obvious even without identifying the individual. Its also susceptible to background knowledge and linkage attacks where an attacker uses outside information or cross-references with other datasets to re-identify someone, and l-diversity doesnt fully solve this either when the diverse values are semantically similar.

5. Differential privacy guarantees that the output of a query on the dataset would look essentially the same whether your personal data was included or not. So an attacker looking at the results cannot determine with any meaningful confidence whether a specific individual's data was part of the dataset, meaning your individual privacy is protected even if the query results are made public.

6. Differential privacy works by adding carefully calibrated random noise to the results of a query before releasing them. The noise is large enough to mask any single individual's contribution, but small enough that the overall aggregate trends in the dataset are still preserved. A simple way to think about it is the randomized response technique: you ask people a sensitive yes/no question, but have them flip a coin first and answer truthfully only if it lands heads, otherwise they give a random answer. This way you can still estimate the true proportion from the aggregate, but no individual response can be traced back to the truth.

7. Some additional approaches include encrypting data at rest and in transit, enforcing strong access controls with least privilege, and requiring multi factor authentication for sensitive systems. We could also strengthen security through regular backups, with copies stored securely and tested through recovery drills. Network segmentation helps limit how far an attacker can move if one system is compromised. Data loss prevention tools can reduce the risk of accidental or unauthorized sharing of sensitive information. Beyond technical controls, organizations can improve security by classifying data based on sensitivity, applying stricter protections to high risk information, and maintaining detailed audit logs for monitoring and incident investigation.

YOUR ANSWERS HERE

1. Differential privacy produces a very different salary for the new employee because it adds random noise to the output in order to protect that individual’s information. Without differential privacy, the result reflects the true value much more directly. However,  with differential privacy, the answer is intentionally distorted so that it becomes harder to infer one person’s exact data. In this case, since the result is strongly influenced by a single employee’s salary, the noisy output can look very different from the non-private one. We observe that this is mainly a feature of differential privacy, because the difference is exactly what helps preserve privacy. However, it also shows the practical downside that the result may become less accurate.

2. The tradeoff between privacy and utility means that stronger privacy protection usually reduces the accuracy or usefulness of the results. In differential privacy, this happens because stronger privacy requires adding more noise. The size of the data matters because noise affects small datasets more strongly than large datasets. In a large dataset, the overall pattern is more stable, so the added noise has less impact on the final result. In a small dataset, each individual contributes more to the outcome, so the noise can change the answer much more noticeably.
